```python
import os
import json
import pandas as pd

m_path = os.path.split(os.getcwd())[0]
data_path = os.path.join(m_path, 'data')

def read_json(json_path):
    with open(json_path) as json_file:
        json_data = json.load(json_file)
    return json_data
```

```python
polymarket_metadata = read_json(os.path.join(m_path, "data/polymarket/metadata.json"))
print(polymarket_metadata)
```

```python
kaggle_smd_path = os.path.join(data_path, "KaggleSMD_01.04.2020", "archive")
metadata = pd.read_csv(os.path.join(kaggle_smd_path, 'symbols_valid_meta.csv'))
metadata
```

```python
stocks_path = os.path.join(kaggle_smd_path, 'stocks')
stocks_dir = os.listdir(stocks_path)
stocks_dir.sort()
ex_stock = pd.read_csv(os.path.join(stocks_path, stocks_dir[0]))
print('Example file:', stocks_dir[0])
ex_stock_name = str(metadata.loc[metadata['Symbol'] == os.path.splitext(stocks_dir[0])[0], 'Security Name'])
ex_stock_name = ex_stock_name.split('Common Stock', 1)[0].split('0', 1)[-1].strip()
print('Company:', ex_stock_name)
ex_stock
```

```python
etfs_path = os.path.join(kaggle_smd_path, 'etfs')
etfs_dir = os.listdir(etfs_path)
etfs_dir.sort()
ex_etf = pd.read_csv(os.path.join(etfs_path, etfs_dir[0]))
print('Example file:', etfs_dir[0])
ex_etf_name = str(metadata.loc[metadata['Symbol'] == os.path.splitext(etfs_dir[0])[0], 'Security Name'])
ex_etf_name = ex_etf_name.split('ETF', 1)[0].split('2', 1)[-1].strip()
print('ETF:', ex_etf_name)
ex_etf
```
