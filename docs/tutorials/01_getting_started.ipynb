# NGLab: Getting Started

Welcome to the NGLab tutorial! This notebook will guide you through the core components of the NGLab simulation engine, which is written in high-performance Rust and exposed via Python bindings.

## 1. Importing the Library

First, let's import the `nglab` module and check that everything is working.

```python
import nglab
print(f"NGLab initialized successfully!")
```

## 2. Using the Order Book

The `OrderBook` is a central component of our market simulation. It handles limit and market orders with microsecond-level latency.

```python
from nglab import OrderBook

# Create a new order book
ob = OrderBook()

# Check initial state
print(f"Spread: {ob.spread()}")
print(f"Mid Price: {ob.mid_price()}")
```

## 3. Polymarket Arena

The `PolymarketArena` simulates prediction market dynamics, allowing you to backtest strategies against historical or synthetic crowd behavior.

```python
from nglab import PolymarketArena

# Initialize arena with $10,000 collateral and 0.1% taker fee
arena = PolymarketArena(initial_collateral=10000.0, taker_fee=0.001)

print(f"Account Value: ${arena.account_value():,.2f}")
```

## 4. Gymnasium Trading Environment

For Reinforcement Learning, we provide a standard Gymnasium-compatible `TradingEnv`.

```python
from nglab import TradingEnv
import numpy as np

# Create environment
env = TradingEnv(initial_capital=10000.0, transaction_cost=0.001)

# Reset environment
obs, info = env.reset(seed=42)
print(f"Initial Observation Shape: {obs.shape}")

# Take a random step (0=Hold, 1=Buy, 2=Sell)
action = 1
next_obs, reward, terminated, truncated, info = env.step(action)
print(f"Step Result - Reward: {reward:.4f}, Done: {terminated}")
```
